```python
# =============================================================================
# 🔧 FIX NUMPY COMPATIBILITY ISSUES
# =============================================================================
import subprocess
import sys

print("""
╔═════════════════════════════════════════════════════════════╗
║              FIXING NUMPY COMPATIBILITY ISSUES              ║
╚═════════════════════════════════════════════════════════════╝
""")

print("📦 Step 1: Uninstalling incompatible packages.. .\n")

# Uninstall packages that have binary incompatibility
packages_to_uninstall = ["pandas", "numpy", "pytesseract", "opencv-python", "pyarrow"]

for package in packages_to_uninstall:
    try:
        subprocess.check_call([sys.executable, "-m", "pip", "uninstall", package, "-y"])
        print(f"✅ Uninstalled {package}")
    except: 
        print(f"⚠️ {package} not found (skipping)")

print("\n" + "=" * 60)
print("📦 Step 2: Installing compatible versions...\n")

# Install packages in the correct order with compatible versions
install_order = [
    "numpy==1.26.4",  # Specific stable NumPy 1.x version
    "pandas==2.1.4",  # Compatible with NumPy 1.26
    "opencv-python==4.9.0.80",
    "pytesseract==0.3.10",
]

for package in install_order: 
    print(f"Installing {package}...")
    try:
        subprocess.check_call([sys.executable, "-m", "pip", "install", package, "--no-cache-dir"])
        print(f"✅ {package} installed\n")
    except Exception as e:
        print(f"❌ Failed:  {e}\n")

print("=" * 60)
print("🔄 Step 3: Verifying installations.. .\n")

# Verify imports
import_tests = {
    "numpy": "import numpy; print(f'NumPy version: {numpy.__version__}')",
    "pandas": "import pandas; print(f'Pandas version: {pandas.__version__}')",
    "opencv": "import cv2; print(f'OpenCV version: {cv2.__version__}')",
    "pytesseract": "import pytesseract; print('Pytesseract OK')",
}

all_ok = True
for name, import_statement in import_tests.items():
    try:
        exec(import_statement)
        print(f"✅ {name} - OK")
    except Exception as e:
        print(f"❌ {name} - FAILED: {e}")
        all_ok = False

print("\n" + "=" * 60)

if all_ok:
    print("✅ All packages are now working correctly!")
    print("\n⚠️ IMPORTANT: Please RESTART your Jupyter kernel now!")
    print("   Kernel → Restart Kernel")
else:
    print("⚠️ Some packages still have issues.")
    print("Please restart your Jupyter kernel and run the verification again.")

print("=" * 60)
```
