```python
import os
import urllib.request
import zipfile

# Download the ZIP
url = 'https://github.com/pwklam/chinese-social-media-scrape-updated/archive/main.zip'
zip_path = '/content/main.zip'
urllib.request.urlretrieve(url, zip_path)

# Extract it
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall('/content')

# Rename the folder
extracted_folder = '/content/chinese-social-media-scrape-updated-main'
target_folder = '/content/chinese-social-media-scrape-updated'
if os.path.exists(extracted_folder):
    os.rename(extracted_folder, target_folder)

# Change to the Douyin scraper directory
os.chdir('/content/chinese-social-media-scrape-updated/duoyin/scrape_chinese_social_media')
print("✅ Repo downloaded and directory set")
```

```python
# Read the file
with open('scrape_douyin_post.py', 'r') as f:
    content = f.read()

# Remove import utils
content = content.replace('import utils\n', '')

# Change headless=False to headless=True
content = content.replace('headless=False', 'headless=True')

# Write back
with open('scrape_douyin_post.py', 'w') as f:
    f.write(content)

print("✅ Scraper file updated")
```
